```python
import pandas as pd
```

```python
df = pd.read_csv("../data/enron_spam_data.csv")
```

```python
df.info()
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

# Count missing values
nan_counts = df.isnull().sum()

# Calculate percentage
nan_percentage = (nan_counts / len(df)) * 100

plt.figure(figsize=(8, 5))

bars = plt.bar(nan_counts.index, nan_counts.values)

# Add percentage labels on top of each bar
for bar, percent in zip(bars, nan_percentage):
    plt.text(
        bar.get_x() + bar.get_width() / 2,   # x-position
        bar.get_height(),                    # y-position
        f"{percent:.2f}%",                   # text
        ha='center',
        va='bottom',
        fontsize=10,
        fontweight='bold'
    )

plt.title("Missing (NaN) Values in Each Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
